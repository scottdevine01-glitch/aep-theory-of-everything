# Anti-Entropic Principle: Interactive Demo
## Theory of Everything Demonstration

This notebook provides an interactive demonstration of the AEP framework across physics, cosmology, and neuroscience.

```python
# Cell 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

print("AEP Theory of Everything - Interactive Demo Loaded!")
```

## Core AEP Principle

```python
# Cell 2: Complexity minimization demo
def aep_complexity_plot():
    theories = ['Too Simple', 'AEP Optimal', 'Overfit']
    theory_comp = [20, 45, 80]
    data_comp = [60, 25, 5]
    total_comp = [t + d for t, d in zip(theory_comp, data_comp)]
    
    x = range(len(theories))
    
    plt.figure(figsize=(10, 6))
    plt.bar(x, theory_comp, label='K(T) - Theory Complexity', alpha=0.7)
    plt.bar(x, data_comp, bottom=theory_comp, label='K(E|T) - Data Complexity', alpha=0.7)
    plt.xticks(x, theories)
    plt.ylabel('Complexity (bits)')
    plt.title('AEP: Reality Minimizes Total Descriptive Complexity')
    plt.legend()
    
    # Highlight minimum
    min_idx = total_comp.index(min(total_comp))
    plt.axvline(x=min_idx, color='red', linestyle='--', alpha=0.8)
    plt.text(min_idx, max(total_comp)*0.9, 'AEP SELECTED', 
             ha='center', color='red', weight='bold')
    
    plt.show()
    
    display(Markdown(f"**AEP selects '{theories[min_idx]}' with minimum total complexity: {min(total_comp)} bits**"))

aep_complexity_plot()
```

## Cosmological Predictions

```python
# Cell 3: Hubble tension resolution
def plot_hubble_tension():
    methods = ['Planck CMB', 'SH0ES Cepheids', 'AEP Prediction']
    h0_values = [67.4, 73.0, 73.63]
    errors = [0.5, 1.0, 0.15]
    
    plt.figure(figsize=(10, 6))
    plt.errorbar(range(len(methods)), h0_values, yerr=errors, 
                 fmt='o', capsize=5, capthick=2, markersize=8)
    plt.xticks(range(len(methods)), methods, rotation=45)
    plt.ylabel('H₀ (km/s/Mpc)')
    plt.title('AEP Resolves Hubble Tension: Matches Direct Measurements')
    plt.grid(True, alpha=0.3)
    
    # Highlight AEP prediction
    plt.annotate('AEP Prediction\nZero free parameters', 
                 xy=(2, 73.63), xytext=(1.5, 75),
                 arrowprops=dict(arrowstyle='->', color='red'),
                 color='red', weight='bold')
    
    plt.tight_layout()
    plt.show()

plot_hubble_tension()
```

## Neural Compression Signatures

```python
# Cell 4: Consciousness metrics
def plot_consciousness_metrics():
    metrics = ['Intrinsic\nDimensionality', 'Predictive\nComplexity', 
               'Information\nIntegration', 'Network\nEfficiency']
    conscious = [18.3, 0.124, 0.67, 0.41]
    unconscious = [23.7, 0.158, 0.52, 0.33]
    
    x = np.arange(len(metrics))
    width = 0.35
    
    plt.figure(figsize=(12, 6))
    plt.bar(x - width/2, conscious, width, label='Conscious State', alpha=0.8)
    plt.bar(x + width/2, unconscious, width, label='Unconscious State', alpha=0.8)
    
    plt.ylabel('Metric Value')
    plt.title('Consciousness as Neural Compression: Optimal Complexity Signatures')
    plt.xticks(x, metrics)
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Add effect sizes
    effect_sizes = [1.45, 1.12, 1.23, 1.08]
    for i, effect in enumerate(effect_sizes):
        plt.text(i, max(conscious[i], unconscious[i]) + 0.1, 
                 f'd={effect}', ha='center', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    display(Markdown("**Conscious states show:**\n- Lower dimensionality (better compression)\n- Better predictability\n- Higher information integration"))

plot_consciousness_metrics()
```

## Live AEP Optimization Demo

```python
# Cell 5: Interactive parameter optimization
def interactive_aep_optimization():
    from ipywidgets import interact, FloatSlider
    
    def optimize_parameters(g_guess, lambda_guess):
        # AEP complexity function
        theory_comp = abs(g_guess * 100) + abs(lambda_guess * 50)
        
        # Distance from AEP-optimal values
        g_optimal, lambda_optimal = 0.024, 0.001833
        data_misfit = ((g_guess - g_optimal)**2 + (lambda_guess - lambda_optimal)**2) * 1000
        
        total_complexity = theory_comp + data_misfit
        
        plt.figure(figsize=(10, 4))
        components = [theory_comp, data_misfit]
        labels = ['Theory Complexity', 'Data Misfit']
        
        plt.bar(labels, components, alpha=0.7)
        plt.ylabel('Complexity Cost')
        plt.title(f'Total AEP Complexity: {total_complexity:.1f} bits\n(g={g_guess:.4f}, λ={lambda_guess:.4f})')
        
        # Show optimal values
        plt.text(0.5, max(components)*0.8, 
                 f'AEP Optimal: g={g_optimal:.4f}, λ={lambda_optimal:.4f}', 
                 ha='center', bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue"))
        
        plt.show()
    
    interact(optimize_parameters,
             g_guess=FloatSlider(min=0.01, max=0.05, step=0.001, value=0.03),
             lambda_guess=FloatSlider(min=0.0005, max=0.005, step=0.0001, value=0.002))

interactive_aep_optimization()
```
